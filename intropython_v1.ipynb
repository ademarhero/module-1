```python
x = int(input("Digite x: ")) #aqui a maquina vai pedir para o usuario digitar um valor para x, e esse valor vai ser convertido para inteiro / eu dei o input de 5
y = x ** 2 + x*2 + 16 #aqui a maquina vai calcular o valor de y com base no valor de x digitado pelo usuario, usando a formula y = x^2 + 2x + 16
print("O valor do y é =", y) #aqui a maquina vai imprimir o valor de y na tela, que é calculado com base no valor de x digitado pelo usuario
```

```python
x = int(input("Digite x: ")) #mesma coisa do exemplo anterior, a maquina vai pedir para o usuario digitar um valor para x, e esse valor vai ser convertido para inteiro
y = int(input("Digite y: ")) #aqui a maquina vai pedir para o usuario digitar um valor para y, e esse valor vai ser convertido para inteiro
resto = x % y #aqui a maquina vai calcular o resto da divisão de x por y, usando o operador % que retorna o resto da divisão inteira / cabe acrescentar que essa intel foi retirada do docs.python.org
print("O resto =", resto) #aqui a maquina vai imprimir o valor do resto na tela, que é calculado com base nos valores de x e y digitados pelo usuario
#COLOQUEI 20 PRA X e 6 PRA Y, O RESTO DE 20/6 É 2, ENTÃO A MAQUINA VAI IMPRIMIR "O RESTO = 2" NA TELA
```
